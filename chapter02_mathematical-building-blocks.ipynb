#### Element-wise operations

```python
def naive_relu(x):
    assert len(x.shape) == 2
    x = x.copy()
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            x[i, j] = max(x[i, j], 0)
    return x
```

```python
def naive_add(x, y):
    assert len(x.shape) == 2
    assert x.shape == y.shape
    x = x.copy()
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            x[i, j] += y[i, j]
    return x
```

```python
def naive_vector_product(x, y):
    assert len(x.shape) == 1
    assert len(y.shape) == 1
    assert x.shape[0] == y.shape[0]
    z = 0.0
    for i in range(x.shape[0]):
        z += x[i] * y[i]
    return z
```

```python
def naive_matrix_vector_product(x, y):
    assert len(x.shape) == 2
    assert len(y.shape) == 1
    assert x.shape[1] == y.shape[0]
    z = np.zeros(x.shape[0])
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            z[i] += x[i, j] * y[j]
    return z
```

```python
def naive_matrix_vector_product(x, y):
    z = np.zeros(x.shape[0])
    for i in range(x.shape[0]):
        z[i] = naive_vector_product(x[i, :], y)
    return z
```

```python
def naive_matrix_product(x, y):
    assert len(x.shape) == 2
    assert len(y.shape) == 2
    assert x.shape[1] == y.shape[0]
    z = np.zeros((x.shape[0], y.shape[1]))
    for i in range(x.shape[0]):
        for j in range(y.shape[1]):
            row_x = x[i, :]
            column_y = y[:, j]
            z[i, j] = naive_vector_product(row_x, column_y)
    return z
```

##### A simple Dense class

```python
import keras
from keras import ops

class NaiveDense:
    def __init__(self, input_size, output_size, activation=None):
        self.activation = activation
        self.W = keras.Variable(
            shape=(input_size, output_size), initializer="uniform"
        )
        self.b = keras.Variable(shape=(output_size,), initializer="zeros")

    def __call__(self, inputs):
        x = ops.matmul(inputs, self.W)
        x = x + self.b
        if self.activation is not None:
            x = self.activation(x)
        return x

    @property
    def weights(self):
        return [self.W, self.b]
```

This code defines a NaiveDense class, which is a custom implementation of a fully connected (dense) layer, similar to what you'd find in a deep learning library like Keras. It's designed to illustrate the fundamental operations within such a layer without relying on Keras's built-in Dense layer.

Here's a breakdown:

import keras and from keras import ops: These lines import the Keras library and its ops module, which provides backend-agnostic tensor operations.

class NaiveDense:: This defines the custom dense layer class.

__init__(self, input_size, output_size, activation=None): This is the constructor for the NaiveDense layer.

input_size: Specifies the number of input features for this layer.
output_size: Specifies the number of output units (neurons) in this layer.
activation: An optional activation function (e.g., relu, softmax) to apply to the output. If None, no activation is applied.
self.W: This creates the weight matrix using keras.Variable. It has a shape of (input_size, output_size) and is initialized with random uniform values. These are the trainable parameters that transform the input.
self.b: This creates the bias vector, also a keras.Variable. It has a shape of (output_size,) and is initialized with zeros. The bias is added to the output of the matrix multiplication.
__call__(self, inputs): This method defines the forward pass of the layer, i.e., how it processes input data.

x = ops.matmul(inputs, self.W): This performs a matrix multiplication of the inputs with the weight matrix self.W. This is the core linear transformation.
x = x + self.b: The bias vector self.b is added to the result of the matrix multiplication. This is an element-wise addition, and broadcasting rules apply if the shapes don't perfectly match.
if self.activation is not None: x = self.activation(x): If an activation function was provided during initialization, it's applied to the result. Activation functions introduce non-linearity into the model.
return x: The processed output of the layer is returned.
@property def weights(self): return [self.W, self.b]: This defines a property named weights that allows you to easily access the trainable parameters of the layer (the weight matrix W and the bias vector b). This is useful when you need to collect all trainable variables from a model, for example, during optimization.

##### A simple Sequential class

```python
class NaiveSequential:
    def __init__(self, layers):
        self.layers = layers

    def __call__(self, inputs):
        x = inputs
        for layer in self.layers:
            x = layer(x)
        return x

    @property
    def weights(self):
        weights = []
        for layer in self.layers:
            weights += layer.weights
        return weights
```

Certainly! This code defines a NaiveSequential class, which is a custom, simplified implementation of a sequential model, much like keras.Sequential. It allows you to stack layers together to create a neural network model.

Here's a breakdown:

class NaiveSequential:: This defines the custom sequential model class.

def __init__(self, layers):: This is the constructor for the NaiveSequential model.

It takes a list of layers as an argument. These layers are expected to be callable objects (like the NaiveDense layers you might define or Keras layers) that can process inputs.
self.layers = layers: It simply stores this list of layers as an attribute of the NaiveSequential instance.
def __call__(self, inputs):: This method defines the forward pass of the entire sequential model.

x = inputs: It initializes x with the initial inputs to the model.
for layer in self.layers:: It then iterates through each layer in the list of layers.
x = layer(x): For each layer, it calls the layer with the current x (which is the output of the previous layer or the initial input). The output of the current layer becomes the input for the next layer.
return x: Finally, it returns the output of the last layer, which is the model's prediction.
@property def weights(self):: This defines a property called weights that allows you to easily collect all trainable weights from all layers within the sequential model.

weights = []: It initializes an empty list to store the weights.
for layer in self.layers:: It iterates through each layer in the model.
weights += layer.weights: It assumes each layer itself has a weights property (or method) that returns its own list of weights, and it extends the weights list with these.
return weights: It returns the comprehensive list of all trainable weights from all layers in the model. This is crucial for optimization algorithms that need to update all model parameters.

That's a great question! Non-linearity is a fundamental concept in neural networks, and it's what allows them to learn complex patterns.

What is Non-linearity?
In the context of activation functions like ReLU, "non-linearity" means that the function's output is not directly proportional to its input. If you were to plot a linear function (like y = mx + b), it would be a straight line. A non-linear function, however, can produce curves, bends, and other more intricate shapes.

For example, the ReLU function f(x) = max(0, x) is non-linear because for x < 0, the output is 0, and for x >= 0, the output is x. This creates a 'bend' or 'kink' at x=0, making it not a straight line across its entire domain.

Why is Non-linearity Important?
If all activation functions in a neural network were linear, then regardless of how many layers you stacked, the entire network would behave just like a single linear layer. Let's say you have two linear layers: y = W1 * x + b1 and z = W2 * y + b2. If you substitute y into the second equation, you get z = W2 * (W1 * x + b1) + b2, which simplifies to z = (W2 * W1) * x + (W2 * b1 + b2). This is still just a linear transformation of x (let W_new = W2 * W1 and b_new = W2 * b1 + b2, then z = W_new * x + b_new).

This means a network with only linear layers could only learn linear relationships between input and output. It would be incapable of modeling anything more complex than a straight line or a simple plane. Most real-world data and tasks, however, involve highly complex, non-linear relationships (e.g., recognizing images, understanding speech, predicting stock prices).

Non-linear activation functions are crucial because they:

Enable Learning Complex Patterns: They introduce the ability for the network to approximate any arbitrary function, allowing it to learn intricate and non-linear mappings from inputs to outputs. Without non-linearity, deep neural networks would be no more powerful than simple linear regression models.
Allow for Deep Architectures: They are what makes "deep" learning possible. By stacking multiple non-linear layers, the network can learn hierarchical representations of the data, extracting progressively more abstract and useful features at each layer.
In essence, non-linearity gives neural networks the expressive power to model the complexities of the real world.

##### A batch generator

```python
import math

class BatchGenerator:
    def __init__(self, images, labels, batch_size=128):
        assert len(images) == len(labels)
        self.index = 0
        self.images = images
        self.labels = labels
        self.batch_size = batch_size
        self.num_batches = math.ceil(len(images) / batch_size)

    def next(self):
        images = self.images[self.index : self.index + self.batch_size]
        labels = self.labels[self.index : self.index + self.batch_size]
        self.index += self.batch_size
        return images, labels
```

#### Running one training step

##### The weight update step

```python
learning_rate = 1e-3

def update_weights(gradients, weights):
    for g, w in zip(gradients, weights):
        w.assign(w - g * learning_rate)
```

```python
from keras import optimizers

optimizer = optimizers.SGD(learning_rate=1e-3)

def update_weights(gradients, weights):
    optimizer.apply_gradients(zip(gradients, weights))
```

```python
%%backend tensorflow
def one_training_step(model, images_batch, labels_batch):
    with tf.GradientTape() as tape:
        predictions = model(images_batch)
        loss = ops.sparse_categorical_crossentropy(labels_batch, predictions)
        average_loss = ops.mean(loss)
    gradients = tape.gradient(average_loss, model.weights)
    update_weights(gradients, model.weights)
    return average_loss
```

#### The full training loop

```python
%%backend tensorflow
def fit(model, images, labels, epochs, batch_size=128):
    for epoch_counter in range(epochs):
        print(f"Epoch {epoch_counter}")
        batch_generator = BatchGenerator(images, labels)
        for batch_counter in range(batch_generator.num_batches):
            images_batch, labels_batch = batch_generator.next()
            loss = one_training_step(model, images_batch, labels_batch)
            if batch_counter % 100 == 0:
                print(f"loss at batch {batch_counter}: {loss:.2f}")
```

```python
%%backend tensorflow
from keras.datasets import mnist

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

train_images = train_images.reshape((60000, 28 * 28))
train_images = train_images.astype("float32") / 255
test_images = test_images.reshape((10000, 28 * 28))
test_images = test_images.astype("float32") / 255

fit(model, train_images, train_labels, epochs=10, batch_size=128)
```
